```python
# 1. Function without arguments and without return
def greet():
    print("Welcome to Python")

greet()

# 2. Function with arguments
def add(a, b):
    print("Addition =", a + b)

add(10, 20)

# 3. Function without arguments but with return
def message():
    return "Python Functions"

print(message())

# 4. Function with arguments and return
def multiply(a, b):
    return a * b

print("Product =", multiply(5, 6))

# 5. Default Argument
def student(name="Mathi"):
    print("Student:", name)

student()
student("Supriya")

# 6. Keyword Arguments
def details(name, age):
    print("Name:", name)
    print("Age:", age)

details(age=18, name="Sumathi")

# 7. Variable Length Arguments
def total_marks(*marks):
    print("Total Marks =", sum(marks))

total_marks(80, 90, 85, 95)

# 8. Lambda Function
square = lambda x: x * x
print("Square =", square(6))

# 9. Recursive Function
def factorial(n):
    if n == 1:
        return 1
    return n * factorial(n - 1)

print("Factorial =", factorial(5))

# 10. Nested Function
def outer():
    print("Outer Function")

    def inner():
        print("Inner Function")

    inner()

outer()
```

```python
# Practical 4 - Lists and List Comprehensions

numbers = [10, 20, 30, 40, 50]

print("Original List:", numbers)

print("First Element:", numbers[0])
print("Last Element:", numbers[-1])


#add a new element at the end 
numbers.append(60)
print("After Append:", numbers)


#insert an element at a specific position 
numbers.insert(2, 25)
print("After Insert:", numbers)


#remove an element from the list
numbers.remove(40)
print("After Remove:", numbers)

print("Length of List:", len(numbers))


#for loop to display each element 
print("List Elements:")
for i in numbers:
    print(i)
    
#square of all elements
square = [x**2 for x in numbers]
print("Squares:", square)

#even numbers 
even = [x for x in numbers if x % 2 == 0]
print("Even Numbers:", even)

#odd numbers 
odd = [x for x in numbers if x % 2 != 0]
print("Odd Numbers:", odd)

#doubling each elements 
double = [x*2 for x in numbers]
print("Doubled List:", double)


#creating nested list(matrix)
matrix = [[1, 2], [3, 4], [5, 6]]
flatten = [item for row in matrix for item in row]
print("Flattened List:", flatten)
```

```python
# Practical 5
# Python Program to Demonstrate Operations on Tuples and Sets

# Creating a tuple
t = (10, 20, 30, 40, 50)

# Displaying the original tuple
print("Original Tuple:", t)

# Accessing the first and last elements
print("First Element:", t[0])
print("Last Element:", t[-1])

# Finding the length of the tuple
print("Length of Tuple:", len(t))

# Tuple slicing
print("Tuple Slice:", t[1:4])

# Counting the occurrence of an element
print("Count of 20:", t.count(20))

# Finding the index of an element
print("Index of 30:", t.index(30))

# Traversing the tuple using a for loop
print("Tuple Elements:")
for i in t:
    print(i)

# Concatenating two tuples
t2 = (60, 70)
new_tuple = t + t2
print("Concatenated Tuple:", new_tuple)

# Creating a set
s = {10, 20, 30, 40}

# Displaying the original set
print("Original Set:", s)

# Adding an element to the set
s.add(50)
print("After Add:", s)

# Removing an element from the set
s.remove(20)
print("After Remove:", s)

# Finding the length of the set
print("Length of Set:", len(s))

# Traversing the set using a for loop
print("Set Elements:")
for i in s:
    print(i)

# Creating another set
s2 = {30, 40, 50, 60}
print("Second Set:", s2)

# Performing set operations
print("Union:", s.union(s2))
print("Intersection:", s.intersection(s2))
print("Difference:", s.difference(s2))
print("Symmetric Difference:", s.symmetric_difference(s2))
```

```python
# Merge and Join DataFrames using Pandas

import pandas as pd

# ----- CONCATENATE OPERATIONS ------------------- (all columns name should be same)

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 1:")
print(df1)

#  Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [201, 202, 203, 205],
    "Marks": [85, 78, 80, 65]
}, index=[4, 5, 6, 7])   

print("\nDataFrame 2:")
print(df2)

frames = [df1,df2]
result = pd.concat(frames)
print("concatenate dataframes")
print(result)

# ---------------- MERGE OPERATIONS ---------------- ( at least one column should be same, others can be different  )

# Creating First DataFrame
df1 = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Sakshi", "Rahul", "Ankit", "Priya"]
})

print("DataFrame 1:")
print(df1)

# Creating Second DataFrame
df2 = pd.DataFrame({
    "ID": [101, 102, 103, 105],
    "Marks": [95, 88, 90, 85]
}, index=[0, 1, 2, 3])   

print("\nDataFrame 2:")
print(df2)

# Inner Merge
inner_merge = pd.merge(df1, df2, on="ID", how="inner")

print("\nInner Merge:")
print(inner_merge)

# Left Merge
left_merge = pd.merge(df1, df2, on="ID", how="left")

print("\nLeft Merge:")
print(left_merge)

# Right Merge
right_merge = pd.merge(df1, df2, on="ID", how="right")

print("\nRight Merge:")
print(right_merge)

# Outer Merge
outer_merge = pd.merge(df1, df2, on="ID", how="outer")

print("\nOuter Merge:")
print(outer_merge)

# ---------------- JOIN OPERATIONS ---------------- ( columns names in both datframes is different )

# Creating DataFrames for Join
df3 = pd.DataFrame({
    "Name": ["Amit", "Riya", "Karan"]
}, index=[1, 2, 3])

df4 = pd.DataFrame({
    "Salary": [30000, 40000, 50000]
}, index=[1, 2, 3])

print("\nDataFrame 3:")
print(df3)

print("\nDataFrame 4:")
print(df4)

# Joining DataFrames
joined_df = df3.join(df4)

print("\nJoined DataFrame:")
print(joined_df)

# Joining with Different Join Types
left_join = df3.join(df4, how="left")

print("\nLeft Join:")
print(left_join)

right_join = df3.join(df4, how="right")

print("\nRight Join:")
print(right_join)

outer_join = df3.join(df4, how="outer")

print("\nOuter Join:")
print(outer_join)
```

```python
# PRACTICAL 18
# Basic Data Cleaning and Preprocessing

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\Users\Acer\OneDrive\Desktop\weekly task\students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display original dataset
print("========== ORIGINAL DATASET ==========")
print(df)

# 3. Display dataset information
print("\n========== ORIGINAL INFORMATION ==========")
df.info()

# 4. Check missing values
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())

# 5. Fill missing Age values with mean
if "Age" in df.columns:
    df["Age"] = df["Age"].fillna(df["Age"].mean())

# 6. Fill missing Marks values with mean
if "Marks" in df.columns:
    df["Marks"] = df["Marks"].fillna(df["Marks"].mean())

# 7. Fill missing City values with "Unknown"
if "City" in df.columns:
    df["City"] = df["City"].fillna("Unknown")

# 8. Remove duplicate records
df = df.drop_duplicates()

print("\n========== AFTER REMOVING DUPLICATES ==========")
print(df)

# 9. Remove extra spaces from Name column
if "Name" in df.columns:
    df["Name"] = df["Name"].str.strip()

# 10. Convert Name column to lowercase
if "Name" in df.columns:
    df["Name"] = df["Name"].str.lower()

# 11. Convert City column to lowercase
if "City" in df.columns:
    df["City"] = df["City"].str.lower()

# 12. Convert Age into integer
if "Age" in df.columns:
    df["Age"] = df["Age"].astype(int)

# 13. Rename Marks column as Score
if "Marks" in df.columns:
    df.rename(columns={"Marks": "Score"}, inplace=True)

# 14. Sort dataset by Score
if "Score" in df.columns:
    df = df.sort_values(by="Score", ascending=False)

# 15. Reset the index
df = df.reset_index(drop=True)

# 16. Display cleaned dataset
print("\n========== CLEANED DATASET ==========")
print(df)

# 17. Check missing values after cleaning
print("\n========== MISSING VALUES AFTER CLEANING ==========")
print(df.isnull().sum())

# 18. Display final dataset information
print("\n========== FINAL DATASET INFORMATION ==========")
df.info()
```

```python
#19th
# PRACTICAL 19
# Calculate Basic Statistics from a Dataset

import pandas as pd
import numpy as np

# Loading CSV File
file_path = r"C:\Users\Acer\OneDrive\Desktop\weekly task\students.csv"


# 1. Load the CSV dataset
#df = pd.read_csv(file_path)
df = pd.read_csv(file_path, sep=",", quoting=csv.QUOTE_NONE)

# 2. Display the dataset
print("========== DATASET ==========")
print(df)

# 3. Select Marks column
marks = df["Marks"]

# 4. Remove missing values
marks = marks.dropna()

# 5. Count number of observations
count = marks.count()

# 6. Calculate total marks
total = marks.sum()

# 7. Calculate Mean
mean = marks.mean()

# 8. Calculate Median
median = marks.median()

# 9. Calculate Mode
mode = marks.mode()

# 10. Calculate Standard Deviation
std = marks.std()

# 11. Calculate Variance
variance = marks.var()

# 12. Find Minimum
minimum = marks.min()

# 13. Find Maximum
maximum = marks.max()

# 14. Calculate Range
data_range = maximum - minimum

# 15. Calculate First Quartile
q1 = marks.quantile(0.25)

# 16. Calculate Third Quartile
q3 = marks.quantile(0.75)

# 17. Display results
print("\n========== BASIC STATISTICS ==========")

print("Number of Observations :", count)
print("Total Marks            :", total)
print("Mean                   :", mean)
print("Median                 :", median)
print("Mode                   :", mode.tolist())
print("Standard Deviation     :", std)
print("Variance               :", variance)
print("Minimum                :", minimum)
print("Maximum                :", maximum)
print("Range                  :", data_range)
print("First Quartile (Q1)    :", q1)
print("Third Quartile (Q3)    :", q3)
```

```python
# Data Visualization using Seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Creating Sample Dataset
data = {
    "Maths": [85, 90, 78, 92, 88],
    "Science": [80, 85, 75, 95, 89],
    "English": [78, 83, 72, 90, 84]
}

# Creating DataFrame
df = pd.DataFrame(data)

print("Dataset:")
print(df)

# -------------------- HEATMAP --------------------
sns.heatmap(df, annot=True, cmap="coolwarm")

plt.title("Student Marks Heatmap")
plt.show()

#--------------------BOXPLOT---------------------

#..........BOXPLOT...........
plt.figure(figsize=(5,4))

sns.boxplot(data=df)

plt.title("Boxplot")

plt.show()


#-----------PAIRPLOT-------------
sns.pairplot(df)
plt.show()
```
